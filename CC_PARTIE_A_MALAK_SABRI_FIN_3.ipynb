```python
# --- Fonctions utilitaires pour les schémas (Q2) : à exécuter une fois ---
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import textwrap
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier
from sklearn.mixture import GaussianMixture
from sklearn.decomposition import PCA
def flux(titre, etapes, note=None, couleurs=None):
    """Diagramme de flux horizontal : une boîte par étape, flèches entre les boîtes."""
    n = len(etapes)
    fig, ax = plt.subplots(figsize=(2.3*n+1, 2.6)); ax.axis('off'); ax.set_xlim(0, n); ax.set_ylim(0, 1)
    cols = couleurs or ['#dbeafe', '#fde68a', '#bbf7d0', '#fecaca', '#e9d5ff', '#fed7aa']
    for i, e in enumerate(etapes):
        ax.add_patch(FancyBboxPatch((i+.06, .25), .78, .5, boxstyle='round,pad=0.02', fc=cols[i % len(cols)], ec='k'))
        ax.text(i+.45, .5, textwrap.fill(e, 16), ha='center', va='center', fontsize=9)
        if i < n-1: ax.annotate('', xy=(i+1.06, .5), xytext=(i+.86, .5), arrowprops=dict(arrowstyle='->', lw=1.5))
    ax.set_title(titre, fontsize=11, weight='bold')
    if note: ax.text(n/2, .05, note, ha='center', fontsize=8, style='italic')
    plt.show()
```

## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression linéaire (moindres carrés ordinaires) — *Ordinary Least Squares, OLS*. **Famille :** régression supervisée ; objectif : prédire une valeur numérique.
- **Entrée :** matrice $X$ ($n$ lignes, $p$ variables) et cible $y$. **Apprentissage :** on cherche l'hyperplan qui minimise la somme des carrés des erreurs. **Sortie :** une prédiction $\hat y$ pour toute nouvelle observation.
- **Modèle :** $\hat y_i=\beta_0+\sum_{j=1}^{p}\beta_j x_{ij}$ ; **critère :** $J(\beta)=\frac{1}{2n}\sum_{i=1}^n (y_i-\hat y_i)^2$.
- **Symboles :** $n$ = nombre d'observations, $p$ = nombre de variables, $x_{ij}$ = valeur de la variable $j$ pour l'observation $i$, $\beta_0$ = constante, $\beta_j$ = coefficient de la variable $j$, $J$ = coût moyen quadratique.
- **Paramètres appris :** $\beta_0,\dots,\beta_p$. **Hyperparamètre :** `fit_intercept` (présence ou non de la constante).
- **Étapes :** (1) construire $X$ avec une constante → (2) calculer les coefficients qui minimisent $J$ → (3) prédire avec $X\beta$.
- **Avantages :** très rapide ; coefficients directement interprétables.
- **Limites :** sensible aux valeurs extrêmes et à la colinéarité ; ne capte pas les relations non linéaires. Les coefficients décrivent des variations conditionnelles, **pas une causalité**.

### Schéma explicatif
**Lecture :** les points bleus sont les données ; la droite rouge est le modèle ; chaque trait gris est un résidu $y_i-\hat y_i$. L'algorithme choisit la droite qui rend la somme des carrés de ces traits la plus petite (ici, projection sur une seule variable pour pouvoir dessiner).

```python
# Q2 — Schéma explicatif (original, Matplotlib)
x = Xr_ent[:, 0]; y = yr_ent
m = LinearRegression().fit(x.reshape(-1, 1), y)
xs = np.linspace(x.min(), x.max(), 50)
plt.figure(figsize=(6, 4)); plt.scatter(x, y, s=12, label='observations')
plt.plot(xs, m.predict(xs.reshape(-1, 1)), 'r', label='droite ajustée')
for xi, yi in list(zip(x, y))[:25]: plt.plot([xi, xi], [yi, m.predict([[xi]])[0]], 'gray', lw=.5)
plt.xlabel('variable x1'); plt.ylabel('cible y'); plt.title('Régression linéaire : minimiser les résidus (traits gris)'); plt.legend(); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (finance/gestion) :** prévoir le chiffre d'affaires mensuel d'une enseigne à partir du budget marketing, du prix moyen et du nombre de points de vente.
- **Cas d'usage 2 :** estimer le prix d'un logement à partir de sa surface et de son nombre de pièces.
- **Données et prérequis :** variables numériques (ou catégorielles encodées), relation approximativement linéaire, peu de colinéarité forte, résidus sans structure.
- **Peu approprié :** prévoir une demande fortement saisonnière ou saturante (relation en S) ; une droite sous-apprendrait.
- **Métriques :** MAE, RMSE, R² (sur un jeu de test indépendant) et analyse des résidus.

```python
from sklearn.linear_model import LinearRegression
modele=LinearRegression().fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python

- **Sortie :** la MAE sur le test, c'est-à-dire l'erreur absolue moyenne exprimée dans l'unité de `y`.
- **Ce qu'on peut conclure :** l'écart moyen entre prédiction et vérité sur ces données simulées ; avec un bruit de 13, une MAE du même ordre de grandeur que ce bruit indique que le modèle capte la partie linéaire du signal.
- **Ce qu'on ne peut pas conclure :** rien sur des données réelles (les données sont générées par une relation linéaire, cas idéal pour ce modèle) ; une seule métrique, sans validation croisée.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression Ridge (régularisation L2) — *Ridge regression*. **Famille :** régression supervisée ; objectif : prédire une valeur numérique en stabilisant les coefficients.
- **Intuition :** on ajoute à l'erreur quadratique une pénalité qui punit les coefficients de grande taille, ce qui réduit la variance du modèle.
- **Critère :** $J(\beta)=\frac1n\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^{p}\beta_j^2$.
- **Symboles :** $X$ = matrice $n\times p$, $y$ = cible, $\beta$ = vecteur des coefficients (la constante n'est pas pénalisée), $\lambda$ = force de régularisation, $\lVert\cdot\rVert_2$ = norme euclidienne.
- **Paramètres appris :** les $\beta_j$. **Hyperparamètre :** `alpha` (le $\lambda$), choisi par validation croisée.
- **Étapes :** standardiser les variables → minimiser erreur + pénalité → prédire.
- **Avantages :** stable en présence de variables corrélées ; solution unique et rapide.
- **Limites :** ne met aucun coefficient exactement à zéro (pas de sélection de variables) ; exige des variables standardisées.

### Schéma explicatif
**Lecture :** chaque courbe est un coefficient. À gauche (alpha faible) on retrouve la régression linéaire ; en allant vers la droite, la pénalité rapproche tous les coefficients de zéro **sans les annuler**.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
alphas = np.logspace(-2, 4, 40)
coefs = [Ridge(alpha=a).fit(StandardScaler().fit_transform(Xr_ent), yr_ent).coef_ for a in alphas]
plt.figure(figsize=(6, 4)); plt.plot(alphas, coefs); plt.xscale('log')
plt.xlabel('alpha (force de la pénalité, échelle log)'); plt.ylabel('valeur des coefficients')
plt.title('Ridge : les coefficients se rétractent vers 0 quand alpha augmente'); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** prévoir un rendement ou un chiffre d'affaires à partir d'indicateurs économiques fortement corrélés (inflation, taux, PIB).
- **Cas d'usage 2 :** modèle de prix avec nombreuses variables proches (surface, nombre de pièces, nombre de chambres).
- **Données et prérequis :** variables numériques standardisées, éventuellement colinéaires ; $p$ pouvant être grand.
- **Peu approprié :** quand on veut **éliminer** des variables inutiles (préférer Lasso) ou quand les relations sont très non linéaires.
- **Métriques :** MAE, RMSE, R² sur validation puis test ; courbe de validation selon `alpha`.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** mêmes données synthétiques ; pipeline `StandardScaler` → `Ridge`. **Hyperparamètre :** `alpha=1.0`, fixé sans réglage.
- **Sortie :** la MAE sur le test.
- **Conclusion possible :** comparer à la régression linéaire (même découpage) : sur des données peu colinéaires, la différence est généralement faible.
- **Limites de la conclusion :** `alpha=1.0` n'a pas été optimisé (il faudrait une validation croisée, par ex. `RidgeCV`) ; données synthétiques ; une seule graine.
- **À faire :** noter la MAE et la comparer à celle de l'algorithme 1.

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression Lasso (régularisation L1) — *Least Absolute Shrinkage and Selection Operator*. **Famille :** régression supervisée avec sélection de variables.
- **Intuition :** la pénalité sur la valeur absolue des coefficients pousse certains d'entre eux **exactement à zéro** : le modèle sélectionne automatiquement les variables utiles.
- **Critère :** $J(\beta)=\frac1{2n}\lVert y-X\beta\rVert_2^2+\lambda\sum_{j=1}^{p}|\beta_j|$.
- **Symboles :** $\beta_j$ = coefficients, $\lambda$ = force de la pénalité L1, $|\cdot|$ = valeur absolue, $n$ = nombre d'observations.
- **Paramètre appris :** les $\beta_j$ (beaucoup valent 0). **Hyperparamètres :** `alpha` ($\lambda$), `max_iter`.
- **Étapes :** standardiser → optimiser par descente de coordonnées → garder les variables à coefficient non nul → prédire.
- **Avantages :** modèle parcimonieux et lisible ; sélection de variables intégrée.
- **Limites :** instable si les variables sont très corrélées (il en choisit une au hasard) ; biais sur les coefficients conservés.

### Schéma explicatif
**Lecture :** contrairement à Ridge, les courbes touchent l'axe zéro et y restent : la variable correspondante est **éliminée** du modèle quand `alpha` augmente.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
Xs = StandardScaler().fit_transform(Xr_ent)
alphas = np.logspace(-2, 1.5, 40)
coefs = [Lasso(alpha=a, max_iter=10000).fit(Xs, yr_ent).coef_ for a in alphas]
plt.figure(figsize=(6, 4)); plt.plot(alphas, coefs); plt.xscale('log')
plt.xlabel('alpha (échelle log)'); plt.ylabel('valeur des coefficients')
plt.title('Lasso : les coefficients atteignent exactement 0 un par un'); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** sélectionner, parmi des dizaines de ratios financiers, ceux qui expliquent la rentabilité d'une entreprise.
- **Cas d'usage 2 :** repérer les gènes ou capteurs réellement informatifs parmi des centaines de variables.
- **Données et prérequis :** variables standardisées ; $p$ grand par rapport à $n$ possible ; on suppose que peu de variables comptent vraiment.
- **Peu approprié :** variables nombreuses, toutes utiles et fortement corrélées (préférer Ridge ou Elastic Net).
- **Métriques :** MAE, RMSE, R², nombre de coefficients non nuls ; stabilité de la sélection entre échantillons.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques de régression ; pipeline `StandardScaler` → `Lasso`. **Hyperparamètres :** `alpha=0.1`, `max_iter=10000`.
- **Sortie :** le R² sur le test (part de variance de `y` expliquée par rapport à la moyenne).
- **Conclusion possible :** un R² proche de 1 signale un bon ajustement sur ces données simulées.
- **Limites :** le R² seul ne dit rien des coefficients éliminés (afficher `modele[-1].coef_` pour les voir) ; `alpha` non validé ; données synthétiques.
- **À faire :** afficher les coefficients du Lasso et noter lesquels sont nuls.

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression polynomiale — *Polynomial regression*. **Famille :** régression supervisée ; objectif : modéliser des relations courbes avec un modèle linéaire en ses paramètres.
- **Intuition :** on enrichit les variables avec leurs puissances et produits ($x$, $x^2$, $x_1x_2$…), puis on applique une régression linéaire (souvent pénalisée) sur ces nouvelles colonnes.
- **Modèle :** $\hat y=\beta_0+\beta_1x+\beta_2x^2+\dots+\beta_dx^d$ (cas à une variable).
- **Symboles :** $d$ = degré du polynôme, $\beta_k$ = coefficient du terme de degré $k$, $x$ = variable explicative.
- **Paramètres appris :** les $\beta_k$. **Hyperparamètres :** `degree` (degré) et la pénalité (`alpha` de Ridge).
- **Étapes :** standardiser → créer les termes polynomiaux → ajuster Ridge → prédire.
- **Avantages :** capte des courbures simples ; reste un modèle linéaire facile à ajuster.
- **Limites :** le nombre de colonnes explose avec le degré et $p$ ; fort risque de surapprentissage et mauvaise extrapolation hors du domaine.

### Schéma explicatif
**Lecture :** degré 1 = droite trop rigide (sous-apprentissage) ; degré 3 suit bien la courbe ; degré 12 oscille pour coller au bruit (surapprentissage). Le degré est donc un hyperparamètre à valider.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from sklearn.preprocessing import PolynomialFeatures
rng = np.random.default_rng(42); x = np.sort(rng.uniform(-3, 3, 40)); y = np.sin(x) + rng.normal(0, .25, 40)
xs = np.linspace(-3, 3, 200); plt.figure(figsize=(6, 4)); plt.scatter(x, y, s=14, label='données (démo)')
for d, c in [(1, 'orange'), (3, 'green'), (12, 'red')]:
    m = make_pipeline(PolynomialFeatures(d), LinearRegression()).fit(x.reshape(-1, 1), y)
    plt.plot(xs, m.predict(xs.reshape(-1, 1)), c, label=f'degré {d}')
plt.ylim(-2, 2); plt.legend(); plt.title('Polynomiale : sous-apprentissage (1) vs surapprentissage (12)'); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (commerce) :** relation prix–demande avec rendements décroissants (la hausse du prix réduit de moins en moins la demande).
- **Cas d'usage 2 :** trajectoire d'un capteur ou courbe de croissance régulière.
- **Données et prérequis :** peu de variables, relation lisse, variables standardisées, degré modéré (2 à 4).
- **Peu approprié :** nombreuses variables (explosion des termes) ou prévision hors de la plage observée.
- **Métriques :** MAE, RMSE, R² sur validation ; comparaison des degrés par validation croisée.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Arbre de régression — *Regression tree (CART)*. **Famille :** régression supervisée non linéaire.
- **Intuition :** l'espace des variables est découpé récursivement par des questions « $x_j\le s$ ? » ; chaque feuille prédit la **moyenne** des cibles qu'elle contient.
- **Critère de découpage :** on choisit $(j,s)$ qui minimise $\sum_{i\in G}(y_i-\bar y_G)^2+\sum_{i\in D}(y_i-\bar y_D)^2$.
- **Symboles :** $j$ = variable testée, $s$ = seuil, $G$ et $D$ = groupes gauche et droit, $\bar y_G,\bar y_D$ = moyennes des cibles dans ces groupes.
- **Paramètres appris :** variables et seuils de chaque nœud, valeurs des feuilles. **Hyperparamètres :** `max_depth`, `min_samples_leaf`.
- **Étapes :** tester tous les seuils → choisir le meilleur découpage → répéter → s'arrêter (profondeur, taille) → prédire la moyenne de la feuille.
- **Avantages :** non linéaire, lisible, sans standardisation.
- **Limites :** instable (un petit changement des données change l'arbre) ; prédictions en escalier, extrapolation impossible.

### Schéma explicatif
**Lecture :** on part du haut ; à chaque nœud on suit la branche gauche si la condition est vraie, droite sinon. `squared_error` mesure la dispersion dans le nœud, `samples` le nombre d'observations, `value` la prédiction (moyenne) de la feuille.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from sklearn.tree import plot_tree
arbre = DecisionTreeRegressor(max_depth=2, random_state=42).fit(Xr_ent, yr_ent)
plt.figure(figsize=(9, 4)); plot_tree(arbre, filled=True, feature_names=[f'x{i}' for i in range(5)], fontsize=8)
plt.title('Arbre de régression (profondeur 2) : question dans chaque nœud, valeur moyenne dans les feuilles'); plt.show()
```

### Contextes d'utilisation

- **Cas d'usage 2 :** estimer un loyer selon quartier et surface avec règles simples à expliquer.
- **Données et prérequis :** variables numériques ou encodées ; aucune standardisation nécessaire ; profondeur limitée.
- **Peu approprié :** relation lisse et linéaire (l'arbre la fait en escalier) ; besoin d'extrapoler.
- **Métriques :** MAE, RMSE, R² ; écart entre erreur d'entraînement et de validation pour détecter le surapprentissage.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Forêt aléatoire pour la régression — *Random Forest Regressor*. **Famille :** régression supervisée par ensemble (bagging).
- **Intuition :** on entraîne beaucoup d'arbres sur des échantillons *bootstrap* différents, avec un sous-ensemble aléatoire de variables à chaque découpage, puis on **moyenne** leurs prédictions.
- **Formule :** $\hat y(x)=\frac1B\sum_{b=1}^{B}T_b(x)$.
- **Symboles :** $B$ = nombre d'arbres, $T_b$ = $b$-ième arbre, $x$ = observation à prédire.
- **Paramètres appris :** la structure de chaque arbre. **Hyperparamètres :** `n_estimators` ($B$), `max_depth`, `max_features`, `min_samples_leaf`.
- **Étapes :** tirer un échantillon bootstrap → entraîner un arbre → répéter $B$ fois → moyenner.
- **Avantages :** robuste, réduit fortement la variance d'un arbre seul ; gère les non-linéarités.
- **Limites :** modèle volumineux et moins interprétable ; ne sait pas extrapoler hors des valeurs vues.

### Schéma explicatif
**Lecture :** de gauche à droite : les données sont rééchantillonnées, chaque échantillon nourrit un arbre, et la prédiction finale est la **moyenne** des prédictions des arbres.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
flux('Forêt aléatoire (régression) : agrégation de B arbres',
     ['Données (n lignes)', 'Échantillons bootstrap 1…B', 'Arbre 1 … Arbre B (variables aléatoires)', 'Prédictions individuelles', 'Moyenne = prédiction finale'],
     note='Chaque arbre voit des données et des variables différentes : leurs erreurs se compensent en moyenne.')
```

- **Données et prérequis :** tableaux mixtes, interactions non linéaires, taille moyenne à grande ; pas de standardisation nécessaire.
- **Peu approprié :** extrapolation de tendance (série croissante) ; contexte exigeant une explication simple ligne par ligne.
- **Métriques :** MAE, RMSE, R² ; erreur *out-of-bag* ; importance des variables (avec prudence).

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Gradient boosting pour la régression — *Gradient Boosting Regressor* (ici `HistGradientBoostingRegressor`). **Famille :** régression supervisée par ensemble séquentiel (boosting).
- **Intuition :** on construit des arbres peu profonds **l'un après l'autre**, chacun corrigeant les erreurs (résidus) des précédents.
- **Formule :** $F_m(x)=F_{m-1}(x)+\eta\,h_m(x)$, où $h_m$ est ajusté sur les résidus $r_i=y_i-F_{m-1}(x_i)$.
- **Symboles :** $F_m$ = modèle après $m$ étapes, $\eta$ = taux d'apprentissage, $h_m$ = petit arbre ajouté à l'étape $m$, $r_i$ = résidu de l'observation $i$.
- **Paramètres appris :** les arbres $h_m$. **Hyperparamètres :** `learning_rate` ($\eta$), `max_iter` (nombre d'arbres), `max_depth`.
- **Étapes :** modèle initial (moyenne) → calcul des résidus → ajustement d'un arbre sur ces résidus → mise à jour → répétition.
- **Avantages :** très performant sur données tabulaires ; capte interactions et non-linéarités.
- **Limites :** réglage sensible, risque de surapprentissage ; entraînement séquentiel, moins lisible.

### Schéma explicatif
**Lecture :** c'est une boucle : on calcule l'erreur restante, on entraîne un petit arbre pour la prédire, on l'ajoute (avec un pas $\eta$) et on recommence.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
flux('Gradient boosting : corriger les résidus étape par étape',
     ['Modèle initial F0 = moyenne de y', 'Résidus r = y − F(x)', 'Petit arbre h sur les résidus', 'F ← F + η·h', 'Répéter max_iter fois'],
     note='Chaque nouvel arbre se concentre sur ce que les précédents ont mal prédit.')
```

### Contextes d'utilisation
- **Cas d'usage 1 (commerce/marketing) :** prévoir les ventes par magasin avec promotions, météo, calendrier.
- **Cas d'usage 2 :** estimer un prix d'assurance ou de crédit.
- **Données et prérequis :** données tabulaires, taille moyenne à grande ; `HistGradientBoosting` gère nativement les valeurs manquantes.
- **Peu approprié :** très petit échantillon (surapprentissage) ou besoin d'explication transparente.
- **Métriques :** MAE, RMSE, R² ; courbe d'erreur selon le nombre d'itérations (arrêt anticipé).

```python
from sklearn.ensemble import HistGradientBoostingRegressor
modele=HistGradientBoostingRegressor(max_iter=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

### Analyse de l'expérience Python
- **Entrées :** données synthétiques ; `HistGradientBoostingRegressor`. **Hyperparamètres :** `max_iter=80`, `random_state=42`, taux d'apprentissage par défaut.
- **Sortie :** la MAE sur le test.
- **Conclusion possible :** comparaison avec les algorithmes 1 à 6 sur le même découpage.
- **Limites :** aucun réglage ; données linéaires et peu nombreuses (280 lignes) : le boosting n'est pas dans son terrain le plus favorable.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression logistique — *Logistic regression*. **Famille :** classification supervisée ; objectif : estimer la probabilité d'appartenir à une classe.
- **Intuition :** un score linéaire est transformé en probabilité par la fonction sigmoïde, puis comparé à un seuil (0,5 par défaut).
- **Formule :** $P(y=1\mid x)=\sigma(z)=\frac1{1+e^{-z}}$ avec $z=\beta_0+\sum_j\beta_jx_j$ ; perte : $-\frac1n\sum_i[y_i\ln p_i+(1-y_i)\ln(1-p_i)]$.
- **Symboles :** $\sigma$ = sigmoïde, $z$ = score linéaire, $\beta_j$ = coefficients, $p_i$ = probabilité prédite pour l'observation $i$.
- **Paramètres appris :** les $\beta_j$. **Hyperparamètres :** `C` (inverse de la régularisation), `penalty`, seuil de décision.
- **Étapes :** standardiser → calculer $z$ → probabilité → minimiser la perte logarithmique → classer selon le seuil.
- **Avantages :** probabilités calibrées, coefficients interprétables (rapports de cotes).
- **Limites :** frontière linéaire ; sensible aux variables non standardisées et aux classes déséquilibrées.

### Schéma explicatif
**Lecture :** un score $z$ très négatif donne une probabilité proche de 0, très positif proche de 1 ; $z=0$ correspond à 50 %. Le seuil de décision peut être déplacé selon le coût des erreurs.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
z = np.linspace(-7, 7, 200)
plt.figure(figsize=(6, 3.5)); plt.plot(z, 1/(1+np.exp(-z)), lw=2)
plt.axhline(.5, color='gray', ls='--'); plt.axvline(0, color='gray', ls='--')
plt.xlabel('score linéaire z'); plt.ylabel('probabilité P(y=1|x)')
plt.title('Fonction sigmoïde : score → probabilité (seuil 0,5 en z = 0)'); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** score de défaut de paiement d'un client de crédit.
- **Cas d'usage 2 (marketing) :** probabilité qu'un client réponde à une campagne.
- **Données et prérequis :** cible binaire ; variables standardisées ; absence de colinéarité extrême.
- **Peu approprié :** frontière de décision très non linéaire (images, interactions complexes) sans ingénierie de variables.
- **Métriques :** précision, rappel, F1, ROC-AUC, PR-AUC (si positifs rares), calibration.

```python
from sklearn.linear_model import LogisticRegression
modele=make_pipeline(StandardScaler(),LogisticRegression(max_iter=1000,random_state=GRAINE)).fit(Xc_ent,yc_ent)
print('F1:',f1_score(yc_test,modele.predict(Xc_test)))
```

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** K plus proches voisins — *k-Nearest Neighbors (KNN)*. **Famille :** classification supervisée (aussi régression), méthode à base d'exemples.

- **Paramètres appris :** aucun (le modèle *mémorise* les données). **Hyperparamètres :** $k$ (`n_neighbors`), distance, pondération.
- **Étapes :** standardiser → calculer les distances → retenir les $k$ plus proches → voter.
- **Avantages :** simple, sans hypothèse sur la forme de la frontière.
- **Limites :** prédiction lente sur grands jeux ; très sensible à l'échelle des variables et à la dimension élevée.

### Schéma explicatif

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Classifieur bayésien naïf — *Naive Bayes* (ici gaussien). **Famille :** classification supervisée probabiliste.
- **Intuition :** on applique le théorème de Bayes en supposant les variables **indépendantes** conditionnellement à la classe ; chaque variable suit une loi normale par classe.
- **Formule :** $\hat y=\arg\max_c\; P(c)\prod_{j=1}^{p}\mathcal N(x_j;\mu_{cj},\sigma_{cj}^2)$.
- **Symboles :** $c$ = classe, $P(c)$ = probabilité a priori de la classe, $\mu_{cj},\sigma_{cj}^2$ = moyenne et variance de la variable $j$ dans la classe $c$.
- **Paramètres appris :** $P(c)$, $\mu_{cj}$, $\sigma^2_{cj}$. **Hyperparamètre :** `var_smoothing`.
- **Étapes :** estimer moyennes/variances par classe → calculer la vraisemblance de $x$ pour chaque classe → multiplier par l'a priori → choisir la classe maximale.
- **Avantages :** très rapide, peu de données requises.
- **Limites :** l'hypothèse d'indépendance est rarement vraie ; probabilités souvent mal calibrées.

### Schéma explicatif
**Lecture :** pour une valeur de $x_0$, la courbe la plus haute indique la classe la plus vraisemblable pour cette variable ; Naive Bayes multiplie ces vraisemblances sur toutes les variables et par l'a priori.

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Arbre de décision — *Decision tree (CART)*. **Famille :** classification supervisée.
- **Intuition :** une suite de questions oui/non sur les variables mène à une feuille qui donne la classe majoritaire.
- **Critère :** impureté de Gini $G=1-\sum_{c}p_c^2$ ; on choisit le découpage qui réduit le plus l'impureté pondérée des deux enfants.
- **Symboles :** $p_c$ = proportion de la classe $c$ dans le nœud.
- **Paramètres appris :** variables et seuils de chaque nœud. **Hyperparamètres :** `max_depth`, `min_samples_leaf`, `criterion`.
- **Étapes :** tester les découpages → retenir le meilleur → répéter récursivement → arrêter → prédire la classe de la feuille.
- **Avantages :** règles lisibles ; aucune standardisation.
- **Limites :** instable, tendance au surapprentissage si profond ; frontières en escaliers.

### Schéma explicatif
**Lecture :** chaque nœud pose une question ; `gini` mesure le mélange des classes (0 = feuille pure) ; `value` donne le nombre d'observations de chaque classe ; la couleur indique la classe majoritaire.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from sklearn.tree import plot_tree
arbre = DecisionTreeClassifier(max_depth=2, random_state=42).fit(Xc_ent, yc_ent)
plt.figure(figsize=(9, 4)); plot_tree(arbre, filled=True, feature_names=[f'x{i}' for i in range(6)], class_names=['0', '1'], fontsize=8)
plt.title('Arbre de décision (profondeur 2) : règles et pureté des feuilles'); plt.show()
```

### Contextes d'utilisation

### Présentation

- **Paramètres appris :** les arbres. **Hyperparamètres :** `learning_rate`, `max_iter`, `max_depth`, `l2_regularization`.
- **Étapes :** score initial → gradient de la perte → arbre sur ce gradient → mise à jour → répétition.
- **Avantages :** souvent le meilleur modèle sur tableaux ; gère interactions et manquants.
- **Limites :** réglage et surapprentissage à surveiller ; faible interprétabilité.

### Schéma explicatif

## 15. AdaBoost

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** AdaBoost — *Adaptive Boosting*. **Famille :** classification supervisée par boosting.
- **Intuition :** on entraîne de petits classifieurs faibles (souches d'arbre) successivement ; après chacun, on **augmente le poids** des observations mal classées pour que le suivant s'y concentre.
- **Formules :** erreur pondérée $\varepsilon_m=\sum_i w_i\,\mathbb 1[h_m(x_i)\ne y_i]$ ; poids du classifieur $\alpha_m=\frac12\ln\frac{1-\varepsilon_m}{\varepsilon_m}$ ; décision $H(x)=\text{signe}\big(\sum_m\alpha_m h_m(x)\big)$.

- **Paramètres appris :** les $h_m$ et les $\alpha_m$. **Hyperparamètres :** `n_estimators`, `learning_rate`, profondeur du classifieur faible.
- **Étapes :** poids égaux → entraîner $h_m$ → calculer $\varepsilon_m$ et $\alpha_m$ → réajuster les poids → répéter → vote pondéré.
- **Avantages :** simple, peu de réglages ; bonne précision avec des classifieurs faibles.
- **Limites :** sensible au bruit et aux valeurs aberrantes (leur poids grandit) ; séquentiel.

### Schéma explicatif

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** K-moyennes — *K-means*. **Famille :** clustering (non supervisé) ; objectif : partitionner les observations en $K$ groupes compacts.
- **Intuition :** chaque groupe est représenté par son centre (centroïde) ; on alterne l'affectation de chaque point au centre le plus proche et le recalcul des centres.
- **Critère :** $\min\sum_{k=1}^{K}\sum_{x_i\in C_k}\lVert x_i-\mu_k\rVert^2$.
- **Symboles :** $K$ = nombre de groupes, $C_k$ = groupe $k$, $\mu_k$ = centroïde du groupe $k$.
- **Paramètres appris :** les centroïdes $\mu_k$ et les affectations. **Hyperparamètres :** $K$ (`n_clusters`), `n_init`, initialisation.
- **Étapes :** initialiser $K$ centres → affecter chaque point au plus proche → recalculer les moyennes → répéter jusqu'à stabilité.
- **Avantages :** simple, rapide, passe à l'échelle.
- **Limites :** $K$ à fixer à l'avance ; ne trouve que des groupes sphériques de taille comparable ; sensible aux valeurs aberrantes.

### Schéma explicatif
**Lecture :** de gauche à droite, on suit l'algorithme : les points sont colorés selon le centre le plus proche (affectation) ; les croix rouges sont les centroïdes, recalculés comme moyennes de leur groupe, jusqu'à stabilité.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
rng = np.random.default_rng(42); cent = X_clu[rng.choice(len(X_clu), 3, replace=False)].copy()
fig, axs = plt.subplots(1, 3, figsize=(12, 3.6))
for ax, titre in zip(axs, ['Itération 0 : centres aléatoires', 'Après 1 itération', 'Après convergence']):
    g = ((X_clu[:, None] - cent[None]) ** 2).sum(2).argmin(1)
    ax.scatter(X_clu[:, 0], X_clu[:, 1], c=g, s=10, cmap='viridis'); ax.scatter(cent[:, 0], cent[:, 1], marker='X', s=160, c='red', ec='k'); ax.set_title(titre, fontsize=9)
    n_it = 1 if 'Après 1' in titre else (50 if 'convergence' in titre else 0)
    for _ in range(n_it):
        g = ((X_clu[:, None] - cent[None]) ** 2).sum(2).argmin(1)
        cent = np.array([X_clu[g == k].mean(0) if (g == k).any() else cent[k] for k in range(3)])
plt.suptitle('K-means : les centroïdes (X rouges) se déplacent vers le centre de leur groupe'); plt.tight_layout(); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** segmentation de clientèle (récence, fréquence, montant) pour adapter les offres.
- **Cas d'usage 2 (commerce international) :** regrouper des pays selon leur profil d'importations.
- **Données et prérequis :** variables numériques standardisées ; $K$ choisi par la méthode du coude ou la silhouette.
- **Peu approprié :** groupes de forme allongée ou imbriquée, groupes de tailles très différentes, forte présence de bruit.
- **Métriques :** silhouette, inertie, stabilité entre graines ; indice de Rand ajusté si des étiquettes de référence existent.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Classification hiérarchique ascendante — *Agglomerative hierarchical clustering*. **Famille :** clustering (non supervisé).
- **Intuition :** chaque point démarre seul ; on **fusionne** à chaque étape les deux groupes les plus proches, ce qui produit un arbre (dendrogramme) qu'on coupe à la hauteur voulue.
- **Critère (lien de Ward) :** fusionner les groupes dont l'union augmente le moins l'inertie intra-groupe : $\Delta(A,B)=\frac{|A||B|}{|A|+|B|}\lVert\mu_A-\mu_B\rVert^2$.
- **Symboles :** $A,B$ = deux groupes, $|A|$ = effectif, $\mu_A$ = centre de $A$.
- **Paramètres appris :** l'arbre de fusions. **Hyperparamètres :** nombre de groupes ou seuil de coupe, type de lien (`ward`, `average`, `complete`, `single`), distance.
- **Étapes :** matrice des distances → fusion des plus proches → mise à jour → dendrogramme → coupe.
- **Avantages :** pas besoin de fixer $K$ d'avance ; dendrogramme interprétable.
- **Limites :** coût au moins quadratique en $n$ ; une fusion ne peut pas être défaite.

### Schéma explicatif
**Lecture :** les feuilles sont les observations ; plus la fusion est haute, plus les groupes fusionnés sont éloignés. Une coupe horizontale à une hauteur donnée fournit les groupes (ici 3 couleurs).

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from scipy.cluster.hierarchy import linkage, dendrogram
Z = linkage(X_clu[:60], 'ward')
plt.figure(figsize=(9, 3.6)); dendrogram(Z, no_labels=True, color_threshold=Z[-3, 2])
plt.ylabel('distance de fusion (Ward)'); plt.title('Dendrogramme (60 premiers points) : une coupe horizontale donne les groupes'); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (commerce international) :** classer des pays ou produits selon leurs flux commerciaux et visualiser la proximité.
- **Cas d'usage 2 :** taxonomie de gènes ou d'espèces.
- **Données et prérequis :** variables standardisées ; $n$ de quelques milliers au plus ; choix du lien et de la distance.
- **Peu approprié :** très grands jeux (mémoire quadratique).
- **Métriques :** silhouette, corrélation cophénétique, stabilité de la coupe.

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** DBSCAN — *Density-Based Spatial Clustering of Applications with Noise*. **Famille :** clustering par densité (non supervisé) avec détection de bruit.
- **Intuition :** un groupe est une zone dense ; les points isolés sont du **bruit** (étiquette −1). Le nombre de groupes n'est pas fixé.
- **Critère :** un point est *central* si au moins `min_samples` points se trouvent dans son voisinage $N_\varepsilon(x)=\{x':\lVert x-x'\rVert\le\varepsilon\}$.
- **Symboles :** $\varepsilon$ = rayon du voisinage, `min_samples` = effectif minimal, $N_\varepsilon(x)$ = voisinage de $x$.
- **Paramètres appris :** l'appartenance aux groupes. **Hyperparamètres :** `eps` ($\varepsilon$), `min_samples`.
- **Étapes :** repérer les points centraux → relier ceux qui sont voisins → rattacher les points frontières → étiqueter le reste comme bruit.
- **Avantages :** groupes de forme quelconque ; détecte le bruit.
- **Limites :** très sensible à `eps` ; peine quand les densités varient.

### Schéma explicatif
**Lecture :** deux groupes en forme de croissant, impossibles à séparer pour K-means, sont retrouvés car chaque groupe est une chaîne de zones denses. Les points de couleur isolée (si présents) sont du bruit.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
Xm, _ = make_moons(300, noise=.07, random_state=42)
lab = DBSCAN(eps=.2, min_samples=5).fit_predict(Xm)
plt.figure(figsize=(5.5, 4)); plt.scatter(Xm[:, 0], Xm[:, 1], c=lab, cmap='viridis', s=12)
plt.title('DBSCAN : deux croissants (démo synthétique) retrouvés par densité'); plt.show()
```

- **Données et prérequis :** distances pertinentes, variables standardisées, densité homogène ; `eps` choisi par graphe des k-distances.
- **Peu approprié :** groupes de densités très différentes, grande dimension.
- **Métriques :** silhouette (hors bruit), part de bruit, stabilité vis-à-vis de `eps`.

```python
from sklearn.cluster import DBSCAN
etiquettes=DBSCAN(eps=.8,min_samples=5).fit_predict(X_clu)
print('Groupes:',set(etiquettes),'Bruit:',(etiquettes==-1).sum())
```

## 20. Mélanges gaussiens

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Mélanges gaussiens — *Gaussian Mixture Models (GMM)*. **Famille :** clustering probabiliste (non supervisé).
- **Intuition :** les données proviennent d'un mélange de $K$ lois normales ; chaque point reçoit une **probabilité d'appartenance** à chaque composante (clustering souple).
- **Formule :** $p(x)=\sum_{k=1}^{K}\pi_k\,\mathcal N(x;\mu_k,\Sigma_k)$.
- **Symboles :** $\pi_k$ = poids de la composante $k$ ($\sum\pi_k=1$), $\mu_k$ = moyenne, $\Sigma_k$ = matrice de covariance.
- **Paramètres appris :** $\pi_k,\mu_k,\Sigma_k$ (par l'algorithme EM). **Hyperparamètres :** `n_components`, `covariance_type`.
- **Étapes (EM) :** initialiser → *E* : calculer les probabilités d'appartenance → *M* : mettre à jour $\pi,\mu,\Sigma$ → répéter jusqu'à convergence.
- **Avantages :** groupes elliptiques, probabilités d'appartenance, choix de $K$ par BIC/AIC.
- **Limites :** sensible à l'initialisation (optimum local) ; hypothèse gaussienne.

### Schéma explicatif
**Lecture :** chaque ellipse rouge représente une loi normale (centre = moyenne, forme = covariance). Un point proche d'une frontière peut appartenir partiellement à deux composantes.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from matplotlib.patches import Ellipse
gmm = GaussianMixture(3, random_state=42).fit(X_clu); lab = gmm.predict(X_clu)
plt.figure(figsize=(5.5, 4.5)); plt.scatter(X_clu[:, 0], X_clu[:, 1], c=lab, s=10, cmap='viridis')
for mu, cov in zip(gmm.means_, gmm.covariances_):
    v, w = np.linalg.eigh(cov); ang = np.degrees(np.arctan2(w[1, 1], w[0, 1]))
    plt.gca().add_patch(Ellipse(mu, 4*np.sqrt(v[1]), 4*np.sqrt(v[0]), angle=ang, fill=False, ec='red', lw=2))
plt.title('Mélange gaussien : une ellipse (2σ) par composante'); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** segmentation souple de clients qui appartiennent à plusieurs profils.
- **Cas d'usage 2 (finance) :** modélisation de régimes de marché (calme/volatil) sur les rendements.
- **Données et prérequis :** variables continues standardisées ; nombre de composantes choisi par BIC/AIC.
- **Peu approprié :** données très non gaussiennes (groupes en croissants) ou grande dimension avec peu de données.
- **Métriques :** BIC/AIC, log-vraisemblance, silhouette, stabilité.

```python
from sklearn.mixture import GaussianMixture
modele=GaussianMixture(n_components=3,random_state=GRAINE).fit(X_clu)
print('Poids:',modele.weights_,'Silhouette:',silhouette_score(X_clu,modele.predict(X_clu)))
```

- **Entrées :** `X_clu` (240 points synthétiques, issus de lois normales, donc cas idéal pour un GMM). **Hyperparamètres :** `n_components=3`, `random_state=42`.
- **Sortie :** les poids $\pi_k$ des trois composantes et la silhouette des affectations dures.
- **Conclusion possible :** des poids voisins de 1/3 sont cohérents avec trois groupes de tailles égales ; comparer la silhouette à celle de K-means.
- **Limites :** $K=3$ imposé ; les probabilités d'appartenance (`predict_proba`) ne sont pas exploitées ici ; données simulées.
- **Observé lors de mon exécution de test :** poids = 1/3 chacun et silhouette ≈ 0,869 (même valeur que K-means et Ward : même partition).

## 21. ACP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Analyse en composantes principales (ACP) — *Principal Component Analysis (PCA)*. **Famille :** réduction de dimension linéaire (non supervisée).
- **Intuition :** trouver de nouveaux axes orthogonaux (composantes) le long desquels les données varient le plus, puis ne garder que les premiers.
- **Formule :** les axes sont les vecteurs propres $v_k$ de la matrice de covariance $\Sigma$ ; part de variance expliquée : $\frac{\lambda_k}{\sum_j\lambda_j}$ ; projection $z=V_q^\top(x-\bar x)$.
- **Symboles :** $\Sigma$ = covariance des variables, $\lambda_k$ = valeur propre (variance sur l'axe $k$), $V_q$ = matrice des $q$ premiers vecteurs propres, $\bar x$ = moyenne.
- **Paramètres appris :** moyenne et axes. **Hyperparamètre :** `n_components` (nombre d'axes).
- **Étapes :** standardiser → calculer covariance → vecteurs/valeurs propres → trier → projeter.
- **Avantages :** déterministe, rapide, quantifie la variance conservée.
- **Limites :** uniquement linéaire ; composantes difficiles à interpréter ; sensible à l'échelle.

### Schéma explicatif
**Lecture :** à gauche, les 178 vins projetés sur deux axes (couleur = cépage, utilisé **seulement pour l'affichage**) ; à droite, la variance expliquée par composante : la courbe rouge indique combien d'axes garder pour conserver, par exemple, 80 % de la variance. Données : `load_wine` (vins réels, jeu UCI).

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from sklearn.datasets import load_wine
Xw = StandardScaler().fit_transform(load_wine().data); yw = load_wine().target
p = PCA(2).fit(Xw); Z = p.transform(Xw)
fig, axs = plt.subplots(1, 2, figsize=(10, 3.8))
axs[0].scatter(Z[:, 0], Z[:, 1], c=yw, s=12, cmap='viridis'); axs[0].set_xlabel(f'CP1 ({p.explained_variance_ratio_[0]:.0%} de variance)'); axs[0].set_ylabel(f'CP2 ({p.explained_variance_ratio_[1]:.0%})'); axs[0].set_title('Projection sur les 2 premières composantes')
pf = PCA().fit(Xw); axs[1].bar(range(1, 14), pf.explained_variance_ratio_); axs[1].plot(range(1, 14), np.cumsum(pf.explained_variance_ratio_), 'r-o'); axs[1].set_xlabel('composante'); axs[1].set_title('Variance expliquée (barres) et cumulée (rouge)')
plt.tight_layout(); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (finance) :** résumer des dizaines d'indicateurs macroéconomiques corrélés en quelques facteurs ; ACP de courbes de taux.
- **Cas d'usage 2 :** compression et visualisation de données chimiques, génomiques ou d'images.
- **Données et prérequis :** variables numériques **standardisées**, relations plutôt linéaires.
- **Peu approprié :** structure non linéaire (spirales) ou variables catégorielles brutes.
- **Métriques :** variance expliquée (cumulée), erreur de reconstruction, éboulis.

```python
from sklearn.decomposition import PCA
X_vin=StandardScaler().fit_transform(load_wine().data)
modele=PCA(n_components=2).fit(X_vin)
projection=modele.transform(X_vin)
print('Variance expliquée:',modele.explained_variance_ratio_)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.show()
```

### Analyse de l'expérience Python
- **Entrées :** `load_wine` (178 vins, 13 mesures chimiques, **données réelles**) standardisées. **Hyperparamètre :** `n_components=2`.
- **Sortie :** les parts de variance expliquée par les 2 composantes et un nuage de points 2D.
- **Conclusion possible :** la somme des deux valeurs affichées indique la part d'information conservée dans ce plan ; elle doit être citée avec le nuage.
- **Limites :** le graphique n'est pas coloré par classe ici : il montre la forme du nuage mais pas la séparation des cépages ; la standardisation est ajustée sur toutes les données (acceptable en exploration non supervisée, pas en modélisation prédictive).
- **Observé lors de mon exécution de test :** CP1 ≈ 36,2 % et CP2 ≈ 19,2 % de variance, soit ≈ 55,4 % cumulés : environ 45 % de l'information est perdue dans ce plan.
- **À faire :** noter les deux variances et leur somme.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** t-SNE — *t-distributed Stochastic Neighbor Embedding*. **Famille :** réduction de dimension non linéaire pour la **visualisation**.
- **Intuition :** on transforme les distances en probabilités de voisinage dans l'espace d'origine, puis on place les points en 2D de façon à reproduire ces voisinages ; on minimise l'écart entre les deux distributions.
- **Critère :** divergence de Kullback-Leibler $\mathrm{KL}(P\Vert Q)=\sum_{i\neq j}p_{ij}\ln\frac{p_{ij}}{q_{ij}}$.
- **Symboles :** $p_{ij}$ = similarité entre $i$ et $j$ en haute dimension (gaussienne), $q_{ij}$ = similarité en 2D (loi de Student).
- **Paramètres appris :** les coordonnées 2D de chaque point. **Hyperparamètres :** `perplexity`, `learning_rate`, nombre d'itérations, graine.
- **Étapes :** calculer $p_{ij}$ → initialiser en 2D → calculer $q_{ij}$ → descendre le gradient de KL → obtenir la carte.
- **Avantages :** révèle des groupes locaux non linéaires.
- **Limites :** distances entre groupes et tailles non interprétables ; non reproductible sans graine ; ne projette pas de nouveaux points.

### Schéma explicatif
**Lecture :** l'algorithme compare les voisinages avant/après et déplace les points 2D jusqu'à ce que les deux soient semblables. La carte finale ne doit être lue qu'en termes de **voisinage local**.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
flux('t-SNE : conserver les voisinages locaux',
     ['Données haute dimension X', 'Similarités p_ij (voisins proches)', 'Points 2D initiaux (aléatoires)', 'Similarités q_ij en 2D', 'Descente de gradient sur KL(P‖Q)'],
     note='Résultat : une carte 2D où les points voisins en haute dimension restent voisins.')
```

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** visualiser des profils de clients ou des représentations de textes en 2D.
- **Cas d'usage 2 :** exploration de données génomiques ou d'images.
- **Données et prérequis :** variables standardisées, souvent après une ACP ; jeu de quelques milliers à dizaines de milliers de points.
- **Peu approprié :** comme étape de prétraitement avant un modèle prédictif, ou pour mesurer des distances entre groupes.
- **Métriques :** conservation des voisinages (*trustworthiness*, rang des voisins), divergence KL ; tester plusieurs `perplexity`.

```python
from sklearn.manifold import TSNE
X_vin=StandardScaler().fit_transform(load_wine().data)
projection=TSNE(n_components=2,perplexity=20,random_state=GRAINE).fit_transform(X_vin)
plt.scatter(projection[:,0],projection[:,1],s=10);plt.title('t-SNE');plt.show()
```

### Analyse de l'expérience Python
- **Entrées :** `load_wine` (178 vins réels) standardisés. **Hyperparamètres :** `perplexity=20`, `random_state=42`, `n_components=2`.
- **Sortie :** un nuage 2D non coloré.
- **Conclusion possible :** on peut juger visuellement si des sous-groupes locaux apparaissent.
- **Limites :** pas de couleur ni de métrique de conservation des voisinages ; la forme change avec la `perplexity` et la graine ; les distances globales ne signifient rien.

## 23. UMAP

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** UMAP — *Uniform Manifold Approximation and Projection*. **Famille :** réduction de dimension non linéaire (visualisation, prétraitement).
- **Intuition :** on construit un graphe pondéré des plus proches voisins dans l'espace d'origine, puis on cherche une disposition en basse dimension dont le graphe est le plus proche possible.
- **Critère :** minimisation d'une entropie croisée entre les poids du graphe d'origine $\mu_{ij}$ et ceux du graphe projeté $\nu_{ij}$.
- **Symboles :** $\mu_{ij}$ = force du lien $i$–$j$ en haute dimension, $\nu_{ij}$ = force en basse dimension.
- **Paramètres appris :** les coordonnées projetées. **Hyperparamètres :** `n_neighbors` (échelle locale), `min_dist` (compacité), `n_components`, graine.
- **Étapes :** graphe de $k$ voisins → poids de lien → initialisation spectrale → optimisation par attraction/répulsion.
- **Avantages :** plus rapide que t-SNE, préserve mieux la structure globale, peut projeter de nouveaux points.
- **Limites :** sensible aux hyperparamètres ; axes non interprétables ; résultats variables selon la graine.

### Schéma explicatif
**Lecture :** les points sont d'abord reliés à leurs voisins ; en 2D, les points reliés sont rapprochés et les autres repoussés, jusqu'à stabilisation.

## 24. Isolation Forest

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Forêt d'isolement — *Isolation Forest*. **Famille :** détection d'anomalies (non supervisée).
- **Intuition :** des découpages aléatoires isolent vite les points rares et éloignés ; un point aberrant est isolé en **peu de coupes**, un point normal en demande beaucoup.
- **Score :** $s(x,n)=2^{-\,E[h(x)]/c(n)}$ ; plus $s$ est proche de 1, plus le point est atypique.
- **Symboles :** $h(x)$ = longueur du chemin pour isoler $x$ dans un arbre, $E[h(x)]$ = moyenne sur la forêt, $c(n)$ = longueur moyenne normalisatrice, $n$ = taille de l'échantillon.
- **Paramètres appris :** les arbres d'isolement. **Hyperparamètres :** `n_estimators`, `contamination` (part attendue d'anomalies), `max_samples`.
- **Étapes :** construire des arbres par coupes aléatoires → mesurer la profondeur d'isolement → moyenner → seuiller selon `contamination`.
- **Avantages :** rapide, sans étiquettes, adapté aux grandes données.
- **Limites :** dépend de `contamination` ; moins efficace pour des anomalies proches des données normales.

### Schéma explicatif
**Lecture :** la couleur donne le score d'anomalie (bleu foncé/bas = très atypique) ; les cercles noirs sont les alertes déclenchées par le seuil. Les deux points ajoutés loin des groupes sont les plus isolés.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
from sklearn.ensemble import IsolationForest
Xan = np.vstack([X_clu, [[9, 9], [-8, -8]]])
iso = IsolationForest(contamination=.02, random_state=42).fit(Xan); sc = iso.decision_function(Xan); pred = iso.predict(Xan)
plt.figure(figsize=(6, 4.5)); s = plt.scatter(Xan[:, 0], Xan[:, 1], c=sc, cmap='RdYlBu', s=14); plt.colorbar(s, label='score (bas = atypique)')
plt.scatter(Xan[pred == -1, 0], Xan[pred == -1, 1], facecolors='none', edgecolors='k', s=90, label='alertes (−1)')
plt.legend(); plt.title('Isolation Forest : les points isolés reçoivent un score bas'); plt.show()
```

- **Données et prérequis :** variables numériques ; fraction d'anomalies faible et estimée ; peu ou pas d'étiquettes fiables.
- **Peu approprié :** anomalies contextuelles ou collectives (une valeur normale seule mais anormale à cette date).
- **Métriques :** qualité des alertes (revue manuelle), précision/rappel **uniquement** si des étiquettes fiables existent, PR-AUC, taux d'alertes.

## 25. Bandits manchots

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Bandits manchots — *Multi-armed bandits*. **Famille :** apprentissage par renforcement (cas à un seul état) ; objectif : maximiser le gain cumulé en choisissant entre $K$ options à gain inconnu.
- **Intuition :** à chaque tour on tire un bras et on observe une récompense ; il faut estimer la valeur de chaque bras tout en exploitant les meilleurs : dilemme **exploration / exploitation**.
- **Formules :** valeur estimée $\hat Q_a=\frac1{N_a}\sum r_t$ ; regret cumulé $R_T=T\mu^*-\sum_t\mu_{a_t}$.
- **Symboles :** $a$ = bras, $N_a$ = nombre de tirages du bras, $r_t$ = récompense, $\mu^*$ = moyenne du meilleur bras, $T$ = nombre de tours.
- **Paramètres appris :** les valeurs estimées $\hat Q_a$. **Hyperparamètres :** stratégie d'exploration (ex. $\varepsilon$, paramètre UCB), horizon $T$.
- **Étapes :** choisir un bras → recevoir la récompense → mettre à jour la moyenne → recommencer.
- **Avantages :** simple ; adapté à l'optimisation en ligne.
- **Limites :** pas d'états ni de contexte (sauf bandits contextuels) ; environnement supposé stable.

### Schéma explicatif
**Lecture :** pour chaque bras, la barre de gauche est la vraie probabilité de gain (inconnue de l'agent) et celle de droite l'estimation obtenue en tirant les bras tour à tour. Plus on tire, plus les barres se rapprochent.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
vraies = np.array([.2, .5, .7]); rng = np.random.default_rng(42)
som = np.zeros(3); cpt = np.zeros(3)
for t in range(500):
    b = t % 3; g = int(rng.random() < vraies[b]); som[b] += g; cpt[b] += 1
est = som / cpt; x = np.arange(3)
plt.figure(figsize=(5.5, 3.8)); plt.bar(x - .18, vraies, .36, label='vraie probabilité (inconnue de l\'agent)'); plt.bar(x + .18, est, .36, label='estimation après 500 tirages')
plt.xticks(x, ['bras 0', 'bras 1', 'bras 2']); plt.ylabel('probabilité de gain'); plt.legend(fontsize=8); plt.title('Bandits : estimer la valeur de chaque bras'); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** test A/B adaptatif de plusieurs versions d'une bannière publicitaire.
- **Cas d'usage 2 :** choix de l'article ou du prix à afficher pour maximiser les clics ou les ventes.
- **Données et prérequis :** récompense observable après chaque choix, options peu nombreuses, environnement relativement stable.
- **Peu approprié :** décisions séquentielles où l'action modifie l'état futur (utiliser Q-learning, SARSA, DQN).
- **Métriques :** récompense cumulée, regret, taux de choix du meilleur bras, variation entre graines.

## 26. Epsilon-greedy

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Stratégie $\varepsilon$-greedy (epsilon-glouton) — *Epsilon-greedy*. **Famille :** renforcement (bandits) ; règle de choix pour équilibrer exploration et exploitation.
- **Intuition :** avec probabilité $1-\varepsilon$ on joue le bras dont la valeur estimée est la meilleure (**exploitation**) ; avec probabilité $\varepsilon$ on joue un bras au hasard (**exploration**).
- **Règle :** $a_t=\begin{cases}\arg\max_a\hat Q_a & \text{avec prob. }1-\varepsilon\\ \text{bras aléatoire} & \text{avec prob. }\varepsilon\end{cases}$ ; mise à jour $\hat Q_a\leftarrow\hat Q_a+\frac1{N_a}(r-\hat Q_a)$.
- **Symboles :** $\varepsilon\in[0,1]$ = probabilité d'explorer, $\hat Q_a$ = valeur estimée, $N_a$ = nombre de tirages, $r$ = récompense.
- **Paramètres appris :** les $\hat Q_a$. **Hyperparamètre :** $\varepsilon$ (éventuellement décroissant).
- **Étapes :** tirer au sort explorer/exploiter → choisir le bras → observer → mettre à jour.
- **Avantages :** très simple, converge vers le meilleur bras.
- **Limites :** l'exploration aléatoire gaspille des essais même quand on sait déjà ; $\varepsilon$ à régler.

### Schéma explicatif
**Lecture :** à gauche, la répartition des tirages (le meilleur bras doit dominer, les autres restent explorés environ 10 % du temps) ; à droite, le gain moyen qui augmente à mesure que l'agent apprend.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
rng = np.random.default_rng(42); vraies = np.array([.2, .5, .7]); som = np.zeros(3); cpt = np.zeros(3); eps = .1; gains = []
for t in range(1000):
    b = int(rng.integers(3)) if rng.random() < eps else int(np.argmax((som + 1) / (cpt + 1)))
    g = int(rng.random() < vraies[b]); som[b] += g; cpt[b] += 1; gains.append(g)
fig, axs = plt.subplots(1, 2, figsize=(10, 3.6))
axs[0].bar(['bras 0', 'bras 1', 'bras 2'], cpt); axs[0].set_title('Nombre de tirages par bras')
axs[1].plot(np.cumsum(gains) / np.arange(1, 1001)); axs[1].set_xlabel('tour'); axs[1].set_title('Gain moyen cumulé'); axs[1].set_ylim(0, 1)
plt.suptitle('ε-greedy (ε = 0,1) : l\'agent finit par jouer surtout le meilleur bras'); plt.tight_layout(); plt.show()
```

### Contextes d'utilisation
- **Cas d'usage 1 (marketing) :** allocation de trafic entre plusieurs pages d'accueil ou publicités.
- **Cas d'usage 2 (gestion) :** sélection dynamique de fournisseurs ou de stratégies de prix testées en ligne.
- **Données et prérequis :** retour immédiat de la récompense ; nombre d'options limité.
- **Peu approprié :** coût élevé de chaque essai exploratoire (santé, sécurité) ou récompenses non stationnaires sans mécanisme d'oubli.
- **Métriques :** récompense cumulée, regret, part de tirages du meilleur bras, variation entre graines.

```python
generateur=np.random.default_rng(GRAINE)
vraies_probas=np.array([.2,.5,.7]);somme=np.zeros(3);comptes=np.zeros(3);epsilon=.1
for t in range(1000):
    bras=int(generateur.integers(3)) if generateur.random()<epsilon else int(np.argmax((somme+1)/(comptes+1)))
    gain=int(generateur.random()<vraies_probas[bras]);somme[bras]+=gain;comptes[bras]+=1
print('Nombre de choix:',comptes,'Valeurs estimées:',somme/np.maximum(comptes,1))
```

### Analyse de l'expérience Python
- **Entrées :** trois bras **simulés** (0,2 / 0,5 / 0,7), 1000 tours. **Hyperparamètre :** `epsilon=0.1`, graine 42.
- **Sortie :** le nombre de tirages par bras et les valeurs estimées.
- **Conclusion possible :** si le bras 2 a été le plus tiré, la stratégie a identifié le meilleur bras ; comparer avec le tirage cyclique (algorithme 25).

- **Observé lors de mon exécution de test :** tirages [25, 49, 926] : le bras 2 domine nettement ; valeurs estimées ≈ [0,20 ; 0,67 ; 0,69] (le bras 1 est surestimé, car il n'a été tiré que 49 fois).

## 27. Q-learning

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Q-learning — *Q-learning*. **Famille :** apprentissage par renforcement, **hors politique** (*off-policy*), tabulaire.
- **Intuition :** l'agent apprend une table $Q(s,a)$ qui estime le gain futur d'une action dans un état, en s'appuyant sur le **meilleur** gain possible à l'état suivant.
- **Formule :** $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\max_{a'}Q(s',a')-Q(s,a)\big]$.
- **Symboles :** $s,s'$ = état courant et suivant, $a$ = action, $r$ = récompense, $\alpha$ = taux d'apprentissage, $\gamma$ = facteur d'actualisation, $\max_{a'}$ = meilleure valeur à l'état suivant.
- **Paramètres appris :** la table $Q$. **Hyperparamètres :** $\alpha$, $\gamma$, $\varepsilon$ (exploration), nombre d'épisodes.
- **Étapes :** observer $s$ → choisir $a$ ($\varepsilon$-greedy) → recevoir $r,s'$ → mettre à jour $Q$ → répéter.
- **Avantages :** simple ; apprend une politique optimale sans modèle de l'environnement.
- **Limites :** table inutilisable si l'espace d'états est grand ; sensible aux hyperparamètres.

### Schéma explicatif
**Lecture :** la boucle se répète à chaque pas : l'agent agit, l'environnement répond par une récompense et un nouvel état, et la table $Q$ est corrigée vers la cible $r+\gamma\max Q(s',\cdot)$.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
flux('Boucle agent–environnement du Q-learning',
     ['Agent : lit l\'état s', 'Action a (ε-greedy sur Q)', 'Environnement : renvoie r et s\'', 'Mise à jour de Q(s,a) avec max Q(s\',·)', 'Nouvel état s ← s\''],
     note='Q-learning utilise le MEILLEUR gain futur, même si l\'action suivante réelle sera exploratoire.')
```

### Contextes d'utilisation
- **Cas d'usage 1 (logistique) :** apprendre une règle de réapprovisionnement ou de routage dans un petit réseau avec états discrets.
- **Cas d'usage 2 :** jeux simples et robotique à états discrets.
- **Données et prérequis :** environnement interactif (simulateur) avec états discrets et peu nombreux ; fonction de récompense définie. Les trajectoires sont **générées par interaction**, ce n'est pas un jeu de données réel.
- **Peu approprié :** états continus ou très nombreux (images) : il faut un approximateur (DQN).
- **Métriques :** récompense cumulée par épisode, taux de réussite, variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False)
Q=np.zeros((env.observation_space.n,env.action_space.n));generateur=np.random.default_rng(GRAINE)
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode)
    for pas in range(100):
        action=int(generateur.integers(env.action_space.n)) if generateur.random()<max(.05,1-episode/2500) else int(np.argmax(Q[etat]))
        suivant,recompense,termine,tronque,_=env.step(action)
        cible=recompense+(0 if termine else .95*np.max(Q[suivant]))
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat=suivant
        if termine or tronque:break
print('Politique:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python
- **Entrées :** environnement public Gymnasium `FrozenLake-v1` (grille 4×4, `is_slippery=False`, donc déterministe), 4000 épisodes. **Hyperparamètres :** $\alpha=0{,}3$, $\gamma=0{,}95$, $\varepsilon$ décroissant de 1 à 0,05.
- **Sortie :** la politique, c'est-à-dire l'action préférée (0=gauche, 1=bas, 2=droite, 3=haut) pour chacun des 16 états.
- **Conclusion possible :** si les actions des états du chemin mènent de l'état 0 à l'état 15, l'agent a appris à atteindre l'objectif.
- **Limites :** on n'évalue ni la récompense cumulée ni le taux de réussite ; une seule graine ; environnement déterministe et minuscule ; il s'agit d'une **simulation** et non d'un jeu de données réel. Les actions des états inutiles (trous, état final) sont sans signification.
- **Observé lors de mon exécution de test :** en suivant les actions des états 0, 4, 8, 9, 13, 14 on obtient le chemin 0 → 4 → 8 → 9 → 13 → 14 → 15, qui contourne les trous (états 5, 7, 11, 12) : la politique atteint l'objectif.

## 28. SARSA

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** SARSA — *State–Action–Reward–State–Action*. **Famille :** renforcement, **sur politique** (*on-policy*), tabulaire.
- **Intuition :** comme Q-learning, mais la cible utilise l'action **réellement choisie** à l'état suivant (y compris exploratoire) : l'agent apprend la valeur de la politique qu'il suit, donc se montre plus prudent.
- **Formule :** $Q(s,a)\leftarrow Q(s,a)+\alpha\big[r+\gamma\,Q(s',a')-Q(s,a)\big]$.
- **Symboles :** $s,a$ = état et action courants, $r$ = récompense, $s'$ = état suivant, $a'$ = action effectivement choisie en $s'$, $\alpha$ = taux d'apprentissage, $\gamma$ = actualisation.
- **Paramètres appris :** la table $Q$. **Hyperparamètres :** $\alpha$, $\gamma$, $\varepsilon$, nombre d'épisodes.
- **Étapes :** choisir $a$ en $s$ → observer $r,s'$ → choisir $a'$ en $s'$ → mettre à jour avec $Q(s',a')$ → $s\leftarrow s'$, $a\leftarrow a'$.
- **Avantages :** apprentissage plus sûr quand l'exploration est risquée.
- **Limites :** converge vers la politique suivie (avec exploration) ; même limites tabulaires que Q-learning.

### Schéma explicatif
**Lecture :** les deux algorithmes ont la même boucle ; la différence tient dans la cible. SARSA tient compte de l'exploration qu'il pratique réellement, Q-learning suppose qu'il jouera ensuite au mieux.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
flux('SARSA vs Q-learning : la cible de mise à jour',
     ['Q-learning : cible = r + γ · max Q(s\',a\')', 'SARSA : cible = r + γ · Q(s\',a\') avec a\' réellement tiré', 'Conséquence : SARSA suit sa propre exploration', 'Résultat : politique plus prudente'],
     note='Même boucle agent–environnement ; seule la cible diffère.')
```

### Contextes d'utilisation
- **Cas d'usage 1 (logistique) :** robot d'entrepôt qui apprend un trajet en évitant des zones dangereuses tout en explorant.
- **Cas d'usage 2 :** gestion de stock simulée où les erreurs d'exploration ont un coût.
- **Données et prérequis :** simulateur interactif à états discrets, comme pour Q-learning ; trajectoires générées par interaction.
- **Peu approprié :** états continus ou très nombreux ; quand on souhaite la politique optimale sans tenir compte de l'exploration (Q-learning).
- **Métriques :** récompense cumulée, taux de réussite, variation entre graines.

```python
import gymnasium as gym
env=gym.make('FrozenLake-v1',is_slippery=False);Q=np.zeros((16,4));generateur=np.random.default_rng(GRAINE)
def choisir(etat,epsilon):
    return int(generateur.integers(4)) if generateur.random()<epsilon else int(np.argmax(Q[etat]))
for episode in range(4000):
    etat,_=env.reset(seed=GRAINE+episode);epsilon=max(.05,1-episode/2500);action=choisir(etat,epsilon)
    for pas in range(100):
        suivant,recompense,termine,tronque,_=env.step(action)
        action_suiv=choisir(suivant,epsilon)
        cible=recompense+(0 if termine else .95*Q[suivant,action_suiv])
        Q[etat,action]+=.3*(cible-Q[etat,action]);etat,action=suivant,action_suiv
        if termine or tronque:break
print('Politique SARSA:',Q.argmax(axis=1));env.close()
```

### Analyse de l'expérience Python
- **Entrées :** `FrozenLake-v1` (4×4, `is_slippery=False`), 4000 épisodes. **Hyperparamètres :** $\alpha=0{,}3$, $\gamma=0{,}95$, $\varepsilon$ décroissant (1 → 0,05).
- **Sortie :** la politique SARSA (16 actions).
- **Conclusion possible :** comparer cette politique à celle du Q-learning ; sur cet environnement **déterministe** elles peuvent coïncider sur le chemin utile.
- **Limites :** pas de mesure de récompense ni de taux de réussite ; une seule graine ; la différence entre on-policy et off-policy ne se voit pas bien quand il n'y a pas de glissade ; simulation, pas de données réelles.
- **Observé lors de mon exécution de test :** sur le chemin utile (états 0, 4, 8, 9, 13, 14) SARSA choisit les mêmes actions que le Q-learning ; les différences portent sur des états non visités (ex. états 1 et 2). Cela confirme qu'en environnement déterministe la différence on-policy / off-policy est peu visible.
- **À faire :** noter la politique ; tester `is_slippery=True` et plusieurs graines pour une vraie comparaison (voir Partie B).

## 29. DQN — cible conceptuelle

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Réseau Q profond — *Deep Q-Network (DQN)*. **Famille :** renforcement profond, hors politique ; remplace la table $Q$ par un **réseau de neurones**.
- **Intuition :** quand les états sont trop nombreux (ou continus), un réseau $Q_\theta(s,a)$ généralise entre états proches. Deux astuces stabilisent l'apprentissage : la **mémoire de rejeu** et le **réseau cible**.
- **Cible de Bellman :** $y=r+\gamma\,(1-d)\max_{a'}Q_{\theta^-}(s',a')$, perte $L(\theta)=\mathbb E\big[(y-Q_\theta(s,a))^2\big]$.
- **Symboles :** $\theta$ = poids du réseau en ligne, $\theta^-$ = poids du réseau cible (copie retardée), $d\in\{0,1\}$ = 1 si l'épisode est terminé, $\gamma$ = actualisation.
- **Paramètres appris :** les poids $\theta$. **Hyperparamètres :** $\gamma$, taux d'apprentissage, taille de mémoire et de lot, fréquence de mise à jour du réseau cible, $\varepsilon$.
- **Étapes :** agir ($\varepsilon$-greedy) → stocker la transition → tirer un lot au hasard → calculer les cibles → descendre le gradient → copier périodiquement $\theta\to\theta^-$.
- **Avantages :** gère des espaces d'états continus et de grande taille.
- **Limites :** instable, gourmand en données et en calcul, nombreux hyperparamètres.
- **Remarque :** dans ce notebook le DQN est **conceptuel** : seule la cible de Bellman est calculée, sans réseau entraîné.

### Schéma explicatif
**Lecture :** les expériences sont stockées puis tirées au hasard (ce qui casse la corrélation entre étapes consécutives) ; le réseau en ligne apprend à s'approcher des cibles calculées avec le réseau cible.

```python
# Q2 — Schéma explicatif (original, Matplotlib)
flux('Architecture DQN',
     ['Environnement → (s, a, r, s\', d)', 'Mémoire de rejeu (stockage)', 'Lot aléatoire de transitions', 'Réseau cible → y = r + γ(1−d) max Q_θ⁻', 'Réseau en ligne : minimiser (y − Q_θ)²'],
     note='Le réseau cible est une copie retardée du réseau en ligne, recopiée toutes les N étapes.')
```

### Contextes d'utilisation
- **Cas d'usage 1 (logistique/gestion) :** pilotage de stocks ou d'un entrepôt avec un grand nombre d'états simulés.
- **Cas d'usage 2 :** jeux vidéo (Atari) ou contrôle de systèmes à partir de capteurs.
- **Données et prérequis :** simulateur rapide, beaucoup d'interactions, bibliothèque d'apprentissage profond ; actions **discrètes**.
- **Peu approprié :** petits problèmes tabulaires (Q-learning suffit) ; applications où l'on ne peut pas faire beaucoup d'essais réels.
- **Métriques :** récompense cumulée moyenne, taux de réussite, variation entre graines, courbes d'apprentissage.

```python
# Calcul de cible DQN, sans entraîner un réseau neuronal complet
recompenses=np.array([0.,1.,-1.]);termine=np.array([False,True,False]);meilleur_Q_suivant=np.array([2.,3.,1.]);gamma=.95
cibles=recompenses+gamma*(~termine)*meilleur_Q_suivant
print('Cibles de Bellman:',cibles)
# Un DQN complet ajoute réseau en ligne, réseau cible et mémoire de rejeu.
```

### Analyse de l'expérience Python
- **Entrées :** trois transitions **inventées** pour la démonstration : récompenses (0, 1, −1), indicateurs de fin (faux, vrai, faux), meilleur $Q$ suivant (2, 3, 1), $\gamma=0{,}95$.
- **Sortie :** les cibles de Bellman : $r+\gamma\,(1-d)\max Q'$ ; pour la 2ᵉ transition (épisode terminé) la cible vaut simplement $r$.
- **Conclusion possible :** le code illustre seulement le calcul de la cible, étape centrale du DQN.
- **Limites :** **aucun réseau n'est entraîné**, aucun environnement n'est utilisé, donc aucune performance ne peut être déduite ; une expérience DQN complète est **facultative** et doit être signalée comme telle.
- **À faire :** recalculer à la main les trois cibles pour vérifier la sortie.

## Q7 / Q8 — Régression (jeu *Diabetes*) : référence vs méthodes avancées
Découpage 60 % entraînement / 20 % validation / 20 % test (graine 42). `alpha` de Ridge et Lasso choisi sur la **validation** ; le **test** n'est évalué qu'une fois par modèle.

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.ensemble import RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import time
X, y = diab.data, diab.target
X_tmp, X_te, y_tmp, y_te = train_test_split(X, y, test_size=.2, random_state=SEED)
X_tr, X_va, y_tr, y_va = train_test_split(X_tmp, y_tmp, test_size=.25, random_state=SEED)   # 0,25 x 0,8 = 20 %
print('tailles train/val/test :', len(X_tr), len(X_va), len(X_te))

def choisir_alpha(Modele, grille):
    scores = {a: mean_absolute_error(y_va, make_pipeline(StandardScaler(), Modele(alpha=a)).fit(X_tr, y_tr).predict(X_va)) for a in grille}
    return min(scores, key=scores.get)
a_ridge = choisir_alpha(Ridge, np.logspace(-3, 3, 13)); a_lasso = choisir_alpha(Lasso, np.logspace(-3, 1, 13))
print('alpha Ridge choisi sur validation :', a_ridge, '| alpha Lasso :', a_lasso)

modeles = {
 'Régression linéaire (référence)': make_pipeline(StandardScaler(), LinearRegression()),
 'Ridge': make_pipeline(StandardScaler(), Ridge(alpha=a_ridge)),
 'Lasso': make_pipeline(StandardScaler(), Lasso(alpha=a_lasso, max_iter=20000)),
 'Forêt aléatoire': RandomForestRegressor(n_estimators=200, min_samples_leaf=5, random_state=SEED),
 'Gradient boosting': HistGradientBoostingRegressor(max_iter=100, learning_rate=.05, random_state=SEED)}
lignes = []
for nom, m in modeles.items():
    t0 = time.time(); m.fit(X_tr, y_tr); dt = time.time() - t0; p = m.predict(X_te)
    lignes.append([nom, mean_absolute_error(y_te, p), np.sqrt(mean_squared_error(y_te, p)), r2_score(y_te, p), dt])
res_reg = pd.DataFrame(lignes, columns=['Modèle', 'MAE', 'RMSE', 'R²', 'Temps d\'ajustement (s)']).round(3)
display(res_reg)
ax = res_reg.set_index('Modèle')[['MAE', 'RMSE']].plot.barh(figsize=(7, 3.5), title='Régression (Diabetes, jeu de test) : erreurs, plus bas = mieux'); plt.tight_layout(); plt.show()
```

### Interprétation (à rédiger après lecture du tableau)
Comparez la référence linéaire aux modèles pénalisés et aux ensembles : l'écart de MAE/RMSE est-il notable ? Les ensembles non linéaires apportent-ils un gain réel sur ces 442 patients ? Discutez coût de calcul (colonne temps), interprétabilité (coefficients vs forêt) et limite : **un seul découpage** → l'écart peut être dû au hasard ; pour conclure, répétez avec validation croisée.

**Observé lors de mon exécution de test (scikit-learn 1.9.1, graine 42) — à vérifier chez vous :** MAE ≈ 42,9 (linéaire) contre 42,7 à 43,5 pour les autres modèles ; R² entre 0,446 et 0,464. Les écarts sont inférieurs à 1 point de MAE : aucun modèle avancé n'apporte de gain net sur ce jeu, et la forêt est la plus lente à ajuster.

## Q7 / Q8 — Classification (jeu *Breast Cancer Wisconsin*) : référence vs méthodes avancées
Classe positive = **maligne** (étiquette 0 dans scikit-learn, qu'on recode en 1). Le rappel est la métrique prioritaire (manquer une tumeur maligne coûte cher). `C` de la régression logistique choisi sur validation.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import precision_score, recall_score, f1_score, roc_auc_score, average_precision_score
Xc, yc = cancer.data, 1 - cancer.target          # 1 = maligne
Xc_tmp, Xc_te, yc_tmp, yc_te = train_test_split(Xc, yc, test_size=.2, stratify=yc, random_state=SEED)
Xc_tr, Xc_va, yc_tr, yc_va = train_test_split(Xc_tmp, yc_tmp, test_size=.25, stratify=yc_tmp, random_state=SEED)
Cs = np.logspace(-3, 2, 11)
C_best = max(Cs, key=lambda C: f1_score(yc_va, make_pipeline(StandardScaler(), LogisticRegression(C=C, max_iter=5000)).fit(Xc_tr, yc_tr).predict(Xc_va)))
print('C choisi sur validation :', C_best)
clfs = {
 'Régression logistique (référence)': make_pipeline(StandardScaler(), LogisticRegression(C=C_best, max_iter=5000)),
 'KNN': make_pipeline(StandardScaler(), KNeighborsClassifier(5)),
 'SVM RBF': make_pipeline(StandardScaler(), SVC(probability=True, random_state=SEED)),
 'Forêt aléatoire': RandomForestClassifier(n_estimators=300, random_state=SEED),
 'Gradient boosting': HistGradientBoostingClassifier(max_iter=150, learning_rate=.05, random_state=SEED)}
try:
    from xgboost import XGBClassifier
    clfs['XGBoost'] = XGBClassifier(n_estimators=150, max_depth=3, learning_rate=.05, random_state=SEED, eval_metric='logloss')
except Exception: print('xgboost indisponible : modèle ignoré')
lignes = []
for nom, m in clfs.items():
    t0 = time.time(); m.fit(Xc_tr, yc_tr); dt = time.time() - t0
    p = m.predict(Xc_te); s = m.predict_proba(Xc_te)[:, 1]
    lignes.append([nom, precision_score(yc_te, p), recall_score(yc_te, p), f1_score(yc_te, p), roc_auc_score(yc_te, s), average_precision_score(yc_te, s), dt])
res_clf = pd.DataFrame(lignes, columns=['Modèle', 'Précision', 'Rappel', 'F1', 'ROC-AUC', 'PR-AUC', 'Temps (s)']).round(3)
display(res_clf)
from sklearn.metrics import ConfusionMatrixDisplay
fig, ax = plt.subplots(figsize=(4, 3.5)); ConfusionMatrixDisplay.from_estimator(clfs['Régression logistique (référence)'], Xc_te, yc_te, display_labels=['bénin', 'malin'], ax=ax); ax.set_title('Matrice de confusion — régression logistique (test)'); plt.show()
```

## Q7 / Q8 — Clustering et réduction de dimension (jeu *Wine*)
Les cépages (étiquettes réelles) servent **seulement à l'évaluation externe** (indice de Rand ajusté) ; les critères internes sont la silhouette et la stabilité (ARI entre graines différentes).

```python
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, adjusted_rand_score
Xv = StandardScaler().fit_transform(vin.data); yv = vin.target      # exploration non supervisée (pas de découpage prédictif)
def stabilite(fabrique, n=5):
    labs = [fabrique(s).fit_predict(Xv) for s in range(n)]
    return np.mean([adjusted_rand_score(labs[i], labs[j]) for i in range(n) for j in range(i+1, n)])
algos = {
 'K-means (référence)': lambda s: KMeans(3, n_init=10, random_state=s),
 'Hiérarchique (Ward)': lambda s: AgglomerativeClustering(3),
 'Mélange gaussien': lambda s: GaussianMixture(3, random_state=s),
 'DBSCAN': lambda s: DBSCAN(eps=2.3, min_samples=5)}
lignes = []
for nom, f in algos.items():
    lab = f(SEED).fit_predict(Xv); masque = lab != -1
    sil = silhouette_score(Xv[masque], lab[masque]) if len(set(lab[masque])) > 1 else np.nan
    lignes.append([nom, len(set(lab) - {-1}), int((lab == -1).sum()), sil, stabilite(f), adjusted_rand_score(yv, lab)])
res_clu = pd.DataFrame(lignes, columns=['Méthode', 'Nb groupes', 'Points de bruit', 'Silhouette (interne)', 'Stabilité (ARI entre graines)', 'ARI vs cépages (externe)']).round(3)
display(res_clu)
```

### Interprétation (à rédiger)
La silhouette est un critère **interne** ; l'ARI vs cépages est **externe** : un bon score interne n'implique pas une bonne correspondance avec les cépages. `eps=2.3` de DBSCAN est un choix arbitraire non optimisé (le dire) ; K-means et le GMM dépendent de l'initialisation (colonne stabilité). Pour une méthode déterministe comme Ward, la stabilité vaut 1 par construction.

**Observé lors de mon exécution de test — à vérifier chez vous :** K-means et le mélange gaussien ont une silhouette ≈ 0,285 et un ARI ≈ 0,897 vs cépages ; DBSCAN affiche la silhouette la plus haute (≈ 0,331) mais ne trouve que 2 groupes avec 42 points de bruit et un ARI externe faible (≈ 0,384) : une silhouette élevée ne garantit pas une bonne correspondance avec les cépages.

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE, trustworthiness
p = PCA().fit(Xv); cum = np.cumsum(p.explained_variance_ratio_)
print('ACP — variance expliquée par CP1, CP2 :', p.explained_variance_ratio_[:2].round(3), '| cumulée 2 CP :', cum[1].round(3), '| nb de CP pour 80 % :', int(np.argmax(cum >= .8)) + 1)
proj = {'ACP': PCA(2).fit_transform(Xv), 't-SNE': TSNE(2, perplexity=20, random_state=SEED).fit_transform(Xv)}
try:
    import umap; proj['UMAP'] = umap.UMAP(n_neighbors=15, random_state=SEED).fit_transform(Xv)
except Exception: print('umap indisponible')
fig, axs = plt.subplots(1, len(proj), figsize=(4.2*len(proj), 3.8))
for ax, (nom, Z) in zip(np.atleast_1d(axs), proj.items()):
    ax.scatter(Z[:, 0], Z[:, 1], c=yv, cmap='viridis', s=12); ax.set_title(f'{nom} — trustworthiness = {trustworthiness(Xv, Z, n_neighbors=10):.3f}', fontsize=9)
plt.suptitle('Wine : projections 2D (couleur = cépage, affichage uniquement)'); plt.tight_layout(); plt.show()
```

Les étiquettes médicales sont fiables, donc précision/rappel sont admissibles ici. La méthode est comparée à une référence minimale : des scores tirés au hasard.

```python
from sklearn.ensemble import IsolationForest
rng = np.random.default_rng(SEED)
benin = np.where(cancer.target == 1)[0]; malin = np.where(cancer.target == 0)[0]
idx = np.r_[benin, rng.choice(malin, 20, replace=False)]
Xa = StandardScaler().fit_transform(cancer.data.iloc[idx]); ya = (cancer.target.iloc[idx].values == 0).astype(int)
print('observations :', len(ya), '| anomalies :', ya.sum(), f'({ya.mean():.1%})')
iso = IsolationForest(n_estimators=300, contamination=ya.mean(), random_state=SEED).fit(Xa)
score = -iso.decision_function(Xa); pred = (iso.predict(Xa) == -1).astype(int)
alea = rng.random(len(ya))
print(pd.DataFrame({'Méthode': ['Isolation Forest', 'Scores aléatoires (référence)'],
    'PR-AUC': [average_precision_score(ya, score), average_precision_score(ya, alea)],
    'ROC-AUC': [roc_auc_score(ya, score), roc_auc_score(ya, alea)]}).round(3).to_string(index=False))
print('Alertes (seuil contamination) — précision :', round(precision_score(ya, pred), 3), '| rappel :', round(recall_score(ya, pred), 3))
```

## Q7 / Q8 — Renforcement : Q-learning vs SARSA (Gymnasium *FrozenLake-v1*, **simulation**)
Environnement **glissant** (`is_slippery=True`) : plus réaliste que la version déterministe. 5 graines, 5000 épisodes d'apprentissage, puis évaluation de la politique **gloutonne** sur 500 épisodes. Il s'agit d'interactions simulées : ce n'est **pas** un jeu de données réel.

```python
import gymnasium as gym
def entrainer(algo, graine, episodes=5000, alpha=.1, gamma=.99):
    env = gym.make('FrozenLake-v1', is_slippery=True); rng = np.random.default_rng(graine); Q = np.zeros((16, 4)); recomp = []
    pol = lambda s, e: int(rng.integers(4)) if rng.random() < e else int(rng.choice(np.flatnonzero(Q[s] == Q[s].max())))
    for ep in range(episodes):
        eps = max(.05, 1 - ep / (episodes * .6)); s, _ = env.reset(seed=int(graine * 100000 + ep)); a = pol(s, eps); tot = 0
        for _ in range(100):
            s2, r, fin, tronc, _ = env.step(a); a2 = pol(s2, eps)
            cible = r + (0 if fin else gamma * (Q[s2].max() if algo == 'Q-learning' else Q[s2, a2]))
            Q[s, a] += alpha * (cible - Q[s, a]); s, a = s2, a2; tot += r
            if fin or tronc: break
        recomp.append(tot)
    env.close(); return Q, np.array(recomp)
def evaluer(Q, graine, n=500):
    env = gym.make('FrozenLake-v1', is_slippery=True); succ = 0
    for ep in range(n):
        s, _ = env.reset(seed=10**6 + int(graine) * 1000 + ep)
        for _ in range(100):
            s, r, fin, tronc, _ = env.step(int(np.argmax(Q[s])))
            if fin or tronc: break
        succ += r
    env.close(); return succ / n
lignes = []; courbes = {}
for algo in ['Q-learning', 'SARSA']:
    taux, cum = [], []
    for g in range(5):
        Q, rec = entrainer(algo, g); taux.append(evaluer(Q, g)); cum.append(rec.sum()); courbes.setdefault(algo, []).append(np.convolve(rec, np.ones(200)/200, 'valid'))
    lignes.append([algo, np.mean(cum), np.std(cum), np.mean(taux), np.std(taux)])
res_rl = pd.DataFrame(lignes, columns=['Algorithme', 'Récompense cumulée (moy.)', 'Récompense cumulée (écart-type entre graines)', 'Taux de réussite (moy.)', 'Taux de réussite (écart-type)']).round(3)
display(res_rl)
for algo, c in courbes.items(): plt.plot(np.mean(c, 0), label=algo)
plt.xlabel('épisode (moyenne glissante 200)'); plt.ylabel('récompense moyenne'); plt.legend(); plt.title('FrozenLake glissant : apprentissage moyen sur 5 graines'); plt.show()
```

### Interprétation (à rédiger)
Comparez les deux algorithmes en tenant compte de l'écart-type **entre graines** : si les intervalles se chevauchent, on ne peut pas affirmer que l'un est meilleur. Rappelez que (i) l'environnement est une simulation, (ii) les hyperparamètres n'ont pas été optimisés, (iii) le **DQN complet est facultatif** et n'a pas été entraîné ici.

**Observé lors de mon exécution de test — à vérifier chez vous :** taux de réussite ≈ 0,741 ± 0,009 (Q-learning) contre ≈ 0,737 ± 0,011 (SARSA) sur 5 graines : les intervalles se chevauchent, on ne peut pas dire que l'un est meilleur.
